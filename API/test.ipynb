```python
import os
import sys

# Obtenez le chemin absolu du répertoire du projet pkmn_db
project_dir = os.path.abspath(os.path.join(os.path.dirname(''), '..'))

# Ajoutez le répertoire du projet au sys.path
if project_dir not in sys.path:
    sys.path.insert(0, project_dir)

print("Contenu du répertoire du projet :", os.listdir(project_dir))
print("sys.path :", sys.path)

# Maintenant, essayez d'importer
try:
    from API.pokemon import Pokemon
    print("Importation réussie !")
except ImportError as e:
    print(f"Erreur d'importation : {e}")
    
    # Si l'erreur persiste, vérifions la structure du dossier API
    api_dir = os.path.join(project_dir, 'API')
    if os.path.exists(api_dir):
        print("Contenu du dossier API :", os.listdir(api_dir))
    else:
        print("Le dossier API n'existe pas dans le répertoire du projet.")
```
